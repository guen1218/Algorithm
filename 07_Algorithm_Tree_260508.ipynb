### 1. 이진 트리(Binary Tree)

```python
class TreeNode:
    def __init__(self, data, left, right):
        self.data = data
        self.left = left
        self.right = right
    def set_left(self, left):
        self.left = left
    def set_right(self, right):
        self.right = right
```

```python
my_tree = TreeNode('A', None, None)
new_node = TreeNode('B', None, None)
my_tree.set_left(new_node)
new_node = TreeNode("C", None, None)
my_tree.set_right(new_node)

cur_node = my_tree.left
cur_node.set_left(TreeNode("D", None, None))
cur_node.set_right(TreeNode("E", None, None))

cur_node = my_tree.right
cur_node.set_left(TreeNode("F", None, None))
cur_node.set_right(TreeNode("G", None, None))
```

```python
# 바이너리 트리 생성
d = TreeNode('D', None, None)
e = TreeNode('E', None, None)
b = TreeNode('B', d, e)

f = TreeNode('F', None, None)
c = TreeNode('C', f, None)

a = TreeNode('A', b, c)

my_binary_tree = a
```

```python
#preorder (선위 순회)
def preorder(node):
    # if node == None: return
    print(node.data, end = ' ')
    if node.left: preorder(node.left)
    if node.right: preorder(node.right)
preorder(my_binary_tree)
```

```python
#inorder (중위 순회)
def inorder(node):
    if node == None: return

    inorder(node.left)
    print(node.data, end = ' ')
    inorder(node.right)

inorder(my_binary_tree)
```

```python
# postorder (후위 순회)
def postorder(node):
    if node == None: return

    postorder(node.left) 
    postorder(node.right)
    print(node.data, end = ' ')

postorder(my_binary_tree)
```

### Binary Search Tree

```python
class STreeNode:
    def __init__(self, key, left, right):
        self.key = key
        self.left = left
        self.right = right
```

```python
# BST 생성
t3 = STreeNode(3, None, None)
t12 = STreeNode(12, None, None)
t7 = STreeNode(7, t3, t12)

t27 = STreeNode(27, None, None)
t31 = STreeNode(31, t27, None)
t26 = STreeNode(26, None, t31)

t18 = STreeNode(18, t7, t26)

my_bst = t18
```

```python
def search_bst(n,key):
    if n == None : return None
    elif key == n.key : return n
    elif key > n.key : return search_bst(n.right, key)
    elif key < n.key : return search_bst(n.left, key)

result = search_bst(my_bst, 12)
if result: print(result.key)
else: print("없는 키입니다")
```

```python
def search_bst_iter(n,key):
    while n != None :
        if key == n.key : return n
        elif key < n.key:
            n = n.left
        else :
            n = n.right
    return None

result = search_bst_iter(my_bst, 12)
```

```python
def insert_bst(root, num):
    new_node = STreeNode(num, None, None)
    
    if root is None:
        return new_node
    
    node = root
    while True:
        parent = node
        
        if num < node.key:
            node = node.left
            if node == None:
                parent.left = new_node
                break
        elif num > node.key:
            node = node.right
            if node == None:
                parent.right = new_node
                break
        else:
            break
            
    return root
```

```python
tryy = int(input("몇번 넣으실래요?"))
root_node = None
for i in range(1, tryy+1):
    num = int(input(f"노드를 입력하시오 {i}번째"))
    root_node = insert_bst(root_node, num)

print("중위순회",end=' ')
def preorder(node):
    if node.left: preorder(node.left)
    print(node.key, end = ' ')
    if node.right: preorder(node.right)
preorder(root_node)
```
